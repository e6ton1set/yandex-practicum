## 1. Описание проекта

TED (от англ technology, education, design — технологии, образование, дизайн) — некоммерческий фонд, проводящий популярные конференции. Специалисты, выступающий на TED работают в самых разных областях и проводят короткие выступления на актуальные социальные, культурные и научные темы. 

В разное время на TED-конференциях выступали математик Бенуа Мандельброт, теоретик искусственного интеллекта Марвин Минский, спортсменка Дана Ньяд и основатель Google Ларри Пейдж. В то же время, в истории TED были неоднозначные и даже скандальные выступления. Например, Рэнди Пауэлл выступил в 2010 году с выступлением о псевдонаучной «вихревой математике», а в 2014 году в конференции TEDMED участвовала Элизабет Холмс — основательница печально известного стартапа Theranos.

__Цель исследования__

Исследовать историю TED-конференций и построить визуализации, с помощью которых сделать выводы по заданным вопросам.

## 2. Ход исследования

### 1. Исследование истории конференций TED

* __В каких странах чаще всего проходили конференции?__

✅ Начало активной деятельности пришлось на 2006 год.
Данные можно разделить на два периода:

- период с 2006 по 2012 с пиком 2011-2012 годах;
- период 2013-2017 год с пиком в 2019 году, что на ~40% превышает предыдущий период.

2.  __Как менялось число конференций в разные годы?__

✅ Чаще всего мероприятия проходили в Соеденённых Штатах Америки (неудивительно, это страна, в которй площадка появилась) 
и в соседних странах - Великобритании и Канаде.

3. __Меняется ли тенденция в изменении количества конференций в зависимости от тематики?__

✅ Независимо от года лидируют три темы (по убыванию): культура, наука и технологии.

### 2. Исследование тематик выступлений

1. __Какие категории выступлений были наиболее популярны?__

✅ Обнаружена связь с популярностью тематик по годам. Это всё те же категории: наука, культура и технологии, однако в другом количественном соотношении (по просмотрам)

2. __Какие категории чаще всего и реже всего вызывают аплодисменты аудитории? Какое из выступлений вызвало наибольшее число аплодисментов?__

✅ Чаще всего хлопают на выступлениях, которые посвящены социальным изменениям, образованию и технологиям, а реже всего, - когда речь идёт о глобальных вопросах и здоровье. 
Наибольшее количество апплодисментов было на выступлении "Политический прогресс делают женщины..."

3. __Какие категории чаще всего и реже всего вызывают смех аудитории? Какое из выступлений оказалось самым смешным?__

✅ Чаще смеются, когда говорят о бизнесе и дизайне, а реже, конгда мероприятия посвящены социальным изменениям и глобальным вопросам.
Самое смешное выступление "Саудовец, индиец и иранец заходят в бар 'Катари'."

4. __Есть ли зависимость между длительностью выступления и количеством просмотров?__

✅ Да, слабая линейная положительная корреляция: с ростом длительности растёт число просмотров.
